# Running this notebook on IQM (Garnet / Emerald)

This notebook builds **standard Qiskit circuits** that are *backend-agnostic* — the
construction and visualisation code below is unchanged from the original and does not
depend on any particular simulator or hardware provider.

To **execute** any circuit `qc` built here on an **IQM** quantum computer
(Garnet / Emerald on IQM Resonance), add:

```python
from iqm_backend import get_backend, run_counts
counts = run_counts(qc, "fake_apollo", shots=2048)   # or "garnet:mock" / "garnet" / "emerald"
```

`run_counts` transpiles the circuit to the **native IQM gate set — PRX (phased-RX) + CZ** —
runs it, and returns a Qiskit counts dict. Use `"fake_apollo"` for a free, offline, no-token
local simulator; switch the target string to `"garnet:mock"`, `"garnet"`, or `"emerald"` to
reach IQM Resonance. See `README.md` in this folder for the full target table and setup.


QuantumCircuit.measure method
Use the measure method to measure a QuantumCircuit.

Examples:

````python
from qiskit import QuantumCircuit
 
qc = QuantumCircuit(5, 5)
qc.x(0)
qc.x(1)
qc.x(4)
qc.measure(range(5), range(5)) #  Measures all qubits into the corresponding clbit.
qc.draw('mpl')
````

````python
from qiskit import QuantumCircuit
 
qc = QuantumCircuit(3, 1)
qc.x([0, 2])
qc.measure(1, 0)    # Measure qubit 1 into the classical bit 0.
qc.draw('mpl')
````

Measure class
The Qiskit Measure class measures the specified qubits.

````python
from qiskit.circuit import Measure
 
#....
qc.append(Measure(), [0], [0])  # measure qubit 0 into clbit 0
qc.draw('mpl')
````

QuantumCircuit.measure_all method
To measure all qubits into the corresponding classical bits, use the measure_all method. By default, this method adds new classical bits in a ClassicalRegister to store these measurements.

````python
from qiskit import QuantumCircuit
 
qc = QuantumCircuit(3, 1)
qc.x([0, 2])
qc.measure_all() # Measure all qubits.
qc.draw('mpl')
````

QuantumCircuit.measure_active method
To measure all qubits that are not idle, use the measure_active method. This method creates a new ClassicalRegister with a size equal to the number of non-idle qubits being measured.

````python
from qiskit import QuantumCircuit
 
qc = QuantumCircuit(3, 1)
qc.x([0, 2])
qc.measure_active() # Measure qubits that are not idle, i.e., qubits 0 and 2.
qc.draw('mpl')
````
